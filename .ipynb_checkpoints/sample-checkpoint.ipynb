# League of Legends Match Data Exploration

### Data Dictionary

| Table & Column | Statistical classification                                                             | Meaning                                                                  |
|---|----------------------------------------------------------------------------------------|--------------------------------------------------------------------------|
| `MatchStatsId`, `MatchId` | ID's                                                                                   | Unique record / match identifier                                         |
| `GameDuration` | Numerical measurement                                                                  | Match length, measured in seconds                                        |
| `TotalGold` | Numerical count                                                                        | Gold earned by the player                                                |
| `DmgDealt`, `DmgTaken` | Numerical count                                                                        | Damage dealt to / taken from champions                                   |
| `MinionsKilled` | Numerical count                                                                        | equals 0 in non-laning modes or for supports.                            |
| `kills`, `deaths`, `assists` | Numerical counts                                                                       | summarizes combat participation.                                         |
| `Lane` | Unordered category                                                                     | `TOP`, `JUNGLE`, `MIDDLE`, `BOTTOM`, `NONE`                              |
| `ChampionFk`, `item1`–`item6` | ID's                                                                                   | Champion ID and Item IDs                                                 |
| `RankFk` / `RankId` | Ordered category  | Competitive rank tier (`0 = Unranked`, `1 = Iron` ... `10 = Challenger`) |
| `Win` | Binary category                                           | `0` = Loss, `1` = Win                                                    |

```python
from pathlib import Path
import kagglehub
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

if 'dataset_path' not in globals():
  path = kagglehub.dataset_download(
      'nathansmallcalder/lol-match-history-and-summoner-data-80k-matches'
  )
  dataset_path = Path(path)

if 'champion_df' not in globals():
  print('loading ChampionTbl.csv...')
  champion_df = pd.read_csv(next(dataset_path.rglob('ChampionTbl.csv')))

if 'rank_df' not in globals():
  print('loading RankTbl.csv...')
  rank_df = pd.read_csv(next(dataset_path.rglob('RankTbl.csv')))

if 'match_df' not in globals():
  print('loading MatchTbl.csv...')
  match_df = pd.read_csv(next(dataset_path.rglob('MatchTbl.csv')))

if 'summoner_match_df' not in globals():
  print('loading SummonerMatchTbl.csv...')
  summoner_match_df = pd.read_csv(
      next(dataset_path.rglob('SummonerMatchTbl.csv'))
  )

if 'match_stats_df' not in globals():
  print('loading MatchStatsTbl.csv (173 MB)...')
  match_stats_df = pd.read_csv(next(dataset_path.rglob('MatchStatsTbl.csv')))

print('all tables loaded')

# connecting tables into one DataFrame 'lol_data'
merged_summoner = pd.merge(
    match_stats_df,
    summoner_match_df[
        ['SummonerMatchId', 'SummonerFk', 'MatchFk', 'ChampionFk']
    ],
    left_on='SummonerMatchFk',
    right_on='SummonerMatchId',
    how='left',
)

lol_data = pd.merge(
    merged_summoner,
    match_df[['MatchId', 'Patch', 'QueueType', 'RankFk', 'GameDuration']],
    left_on='MatchFk',
    right_on='MatchId',
    how='left',
)

lol_data = pd.merge(
    lol_data, champion_df, left_on='ChampionFk', right_on='ChampionId', how='left'
)
lol_data = pd.merge(
    lol_data, rank_df, left_on='RankFk', right_on='RankId', how='left'
)

# seconds -> minutes
lol_data['GameDuration_min'] = (lol_data['GameDuration'] / 60.0).round(2)

display(
    lol_data[[
        'MatchStatsId',
        'TotalGold',
        'DmgDealt',
        'MinionsKilled',
        'Lane',
        'ChampionName',
        'RankName',
        'Win',
    ]].head()
)
```

```python
# Missing values
print("--- Missing values per column (top 10) ---")
display(lol_data.isna().sum().sort_values(ascending=False).head(10))

# count, mean, std, min, quartiles, max
summary_cols = ["TotalGold", "DmgDealt", "DmgTaken", "MinionsKilled", "TurretDmgDealt", "kills", "deaths", "assists", "visionScore", "GameDuration_min"]
print("\n--- Numerical Summary (.describe()) ---")
display(lol_data[summary_cols].describe().round(2))

# comparasion of mean and median for DmgDealt, TotalGold and MinionsKilled
for col in ["DmgDealt", "TotalGold", "MinionsKilled", "visionScore"]:
    mean_val = lol_data[col].mean()
    median_val = lol_data[col].median()
    print(f"{col:15s} -> Mean: {mean_val:10.2f} | Median: {median_val:10.2f} | Diff (Mean - Median): {mean_val - median_val:8.2f}")

# modes and lanes
print("\n--- Most common QueueTypes ---")
display(lol_data["QueueType"].value_counts(dropna=False).head(6))

print("\n--- Most common Lanes (including NaN and NONE) ---")
display(lol_data["Lane"].value_counts(dropna=False))

# 5. anomaly inspection: CurrentMasteryPoints vs ChampionFk and extrerme kill counts or 0 CS
print("\n--- Anomaly Check 1: Does CurrentMasteryPoints equal ChampionFk? ---")
display(lol_data[["MatchStatsId", "ChampionFk", "ChampionName", "CurrentMasteryPoints"]].head(10))
same_count = (lol_data["CurrentMasteryPoints"] == lol_data["ChampionFk"]).sum()
print(f"Records where CurrentMasteryPoints == ChampionFk: {same_count} out of {len(lol_data)}")

print("\n--- Anomaly Check 2: Extreme outlier records (e.g., kills > 35 or GameDuration < 300s / remakes) ---")
display(lol_data.loc[lol_data["kills"] > 35, ["MatchFk", "QueueType", "ChampionName", "kills", "deaths", "DmgDealt", "GameDuration_min"]].head(5))
```

### Exercise 1 — Answers and Explanations

* **Mean vs. Median (`DmgDealt` and `TotalGold`):**
  To describe a typical player's damage dealt (`DmgDealt`) in a match, the **median** is a better summary than the **mean**. In *League of Legends*, damage cannot be negative (it is bounded at 0 on the left), but long games (45+ minutes) or high-kill modes (like ARAM/Arena) produce extreme high-damage outliers (e.g., 60,000–100,000+ damage). These right-tail outliers pull the arithmetic mean significantly above the median (already visible in the first 5 sample rows, where `DmgDealt` has a mean of **16,261.60** but a median of only **8,821.00**, with standard deviation **14,506.82**).
* **Zeros and Missing Values (`Lane`, `visionScore`, and `MinionsKilled`):**
  In `MatchStatsTbl.csv`, `Lane` has **177,309 missing values (24.21%)** and many `"NONE"` entries. Furthermore, `visionScore` and `TurretDmgDealt` have a large spike at `0`. This happens because the dataset mixes different game modes (`CLASSIC` 5v5 Summoner's Rift, `ARAM`, and `Arena`), as well as early surrendered/remade games (`GameDuration < 300s`). In ARAM and Arena, wards do not exist (`visionScore == 0`), so averaging `visionScore` across the entire table underestimates vision control in standard 5v5 matches. A more useful summary is to filter by `QueueType == "CLASSIC"` before summarizing lane-specific or vision metrics.
* **Unusual Values & Data Critique (`CurrentMasteryPoints` and extreme rows):**
  1. Inspecting `CurrentMasteryPoints` reveals a data-engineering bug in the source dataset: the values (`902, 902, 16, 103, 800` in the first 5 rows) match `ChampionFk` from `SummonerMatchTbl.csv`. Treating `CurrentMasteryPoints` as actual mastery experience would lead to completely false conclusions.
  2. Records with `GameDuration < 300` seconds represent `/remake` matches (where a player disconnected at the start), and records with `MinionsKilled == 0` on `TOP` lane (like Row 4 in the sample) occur in special game modes (indicated by Summoner Spells `2201`/`2202`). We need `QueueType` and `GameDuration` context before deciding whether to filter them out.

## Exercise 2 — Explore distributions, relationships, and correlations

**Predict first (Before running the plots):**
1. **Distribution prediction:** Do you expect `DmgDealt` (Damage Dealt) and `TotalGold` to be evenly (uniformly) or symmetrically distributed?
   * *Prediction:* I predict that `DmgDealt` will **not** be evenly distributed; instead, it will be **right-skewed (positively skewed)**. Most matches last 20–30 minutes where players deal moderate damage (10,000–25,000), while a small number of very long or high-action games will create a long tail stretching toward very high values. `TotalGold` should have a more central peak (bell-shaped with a slight right skew, plus a small lower cluster for early surrenders/remakes) because passive gold generation guarantees a baseline amount of gold over time.
2. **Relationship & Correlation prediction:** Do you expect players with more `TotalGold` to deal more `DmgDealt`? Will this trend apply equally to every player?
   * *Prediction:* I expect a **strong positive correlation** between `TotalGold` and `DmgDealt`, because gold buys items that increase attack damage, ability power, and cooldown reduction. However, the spread (variance) will widen at higher gold levels, and the trend will **not** describe every record equally: Tank and Enchanter Support champions (e.g., Soraka, Milio) spend gold on defensive or healing items rather than damage, whereas Poke/Carry champions deal high damage per gold spent.

```python
# Filtrujemy dane do standardowego trybu 5v5 (CLASSIC) bez remake'ów (GameDuration >= 300s),
# aby uniknąć zniekształcenia przez ARAM/Arenę (lub używamy pełnego zbioru jeśli brak MatchTbl)
classic_mask = (lol_data["QueueType"] == "CLASSIC") & (lol_data["GameDuration"] >= 300)
df_classic = lol_data.loc[classic_mask].copy() if classic_mask.sum() > 0 else lol_data.copy()

print(f"Analyzing {len(df_classic):,} valid match records.")

# 1. Histogramy: DmgDealt oraz TotalGold (z zaznaczeniem średniej i mediany)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram DmgDealt
axes[0].hist(df_classic["DmgDealt"], bins=40, color="steelblue", edgecolor="white")
axes[0].axvline(df_classic["DmgDealt"].mean(), color="crimson", linestyle="--", linewidth=2, label=f"Mean: {df_classic['DmgDealt'].mean():,.0f}")
axes[0].axvline(df_classic["DmgDealt"].median(), color="darkorange", linestyle="-", linewidth=2, label=f"Median: {df_classic['DmgDealt'].median():,.0f}")
axes[0].set_xlabel("Total Damage Dealt to Champions")
axes[0].set_ylabel("Number of player match records")
axes[0].set_title("Distribution of Damage Dealt (Right-Skewed)")
axes[0].legend()

# Histogram TotalGold
axes[1].hist(df_classic["TotalGold"], bins=40, color="seagreen", edgecolor="white")
axes[1].axvline(df_classic["TotalGold"].mean(), color="crimson", linestyle="--", linewidth=2, label=f"Mean: {df_classic['TotalGold'].mean():,.0f}")
axes[1].axvline(df_classic["TotalGold"].median(), color="darkorange", linestyle="-", linewidth=2, label=f"Median: {df_classic['TotalGold'].median():,.0f}")
axes[1].set_xlabel("Total Gold Earned")
axes[1].set_ylabel("Number of player match records")
axes[1].set_title("Distribution of Total Gold Earned")
axes[1].legend()

plt.tight_layout()
plt.show()

# 2. Scatterplot: TotalGold vs DmgDealt (próbka 15,000 rekordów dla czytelności przy dużym zagęszczeniu, małe s i alpha)
sample_plot = df_classic.sample(n=min(15000, len(df_classic)), random_state=42)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(sample_plot["TotalGold"], sample_plot["DmgDealt"], s=10, alpha=0.2, color="royalblue")
ax.set_xlabel("Total Gold Earned (Gold)")
ax.set_ylabel("Damage Dealt to Champions")
ax.set_title("Relationship between Total Gold Earned and Damage Dealt")
ax.grid(True, linestyle="--", alpha=0.5)
plt.show()

# 3. Matematyczna analiza korelacji (Pearson vs Spearman) dla zmiennych ilościowych
corr_cols = ["TotalGold", "DmgDealt", "DmgTaken", "MinionsKilled", "TurretDmgDealt", "kills", "deaths", "assists", "visionScore", "Win"]
pearson_corr = df_classic[corr_cols].corr(method="pearson").round(3)
spearman_corr = df_classic[corr_cols].corr(method="spearman").round(3)

print("--- Pearson Linear Correlation Matrix ---")
display(pearson_corr)

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(pearson_corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".2f", ax=ax)
ax.set_title("Pearson Correlation Matrix of Key In-Game Metrics")
plt.show()
```

### Exercise 2 — Post-Plot Verification & Mathematical Explanation

* **Did the results match our prediction? (Distribution):**
  **Yes.** The histogram of `DmgDealt` confirms a strong **right-skewed distribution**. Most observations cluster between 8,000 and 25,000 damage, while extreme games stretch the right tail past 60,000+. Because the mean is sensitive to extreme values while the median only depends on the middle rank (50th percentile), the red dashed line (**mean**) is pulled noticeably to the right of the orange solid line (**median**).
* **Direction, Spread, and Overplotting (Scatterplot):**
  The scatterplot confirms a clear **positive association**: as `TotalGold` increases, `DmgDealt` generally increases. However, the vertical spread (fan shape / heteroscedasticity) widens dramatically as gold increases. A player with **15,000 gold** might deal anywhere from **8,000 damage** (e.g., a tank or support focusing on utility/objectives) to **45,000+ damage** (a hyper-carry mage or marksman). Because there are hundreds of thousands of records, plotting without a small point size (`s=10`) and transparency (`alpha=0.2`) would create a solid blue blob (*overplotting*), hiding the true density of typical games around 9,000–14,000 gold.
* **Mathematical Interpretation of Correlations (Pearson vs. Spearman) & Confounding Limits:**
  * `TotalGold` shows a strong positive correlation with `kills`, `MinionsKilled`, and `DmgDealt`.
  * `deaths` correlates negatively with `Win`, while `TurretDmgDealt` and `TotalGold` correlate positively with `Win`.
  * `visionScore` has a weak or even negative correlation with `MinionsKilled` — not because vision makes you farm worse, but because **Support players** actively place wards (high `visionScore`) while intentionally leaving minion kills (`MinionsKilled`) to their carries!
  * **Why correlation does not prove causation (Confounding variable):** A high correlation between `TotalGold` and `DmgDealt` does not mean that simply giving gold causes linear damage increases in isolation — **Game Duration (`GameDuration`)** is a major *confounding variable* (zmienna zakłócająca). Longer matches naturally allow players to accumulate both more passive/farm gold and more total combat time to deal damage. Additionally, **Champion Role (`Lane`)** determines how gold is converted into stats.

## Exercise 3 — Compare metrics across categories (`RankName` and `Lane`)

`RankFk` (`RankName` in `RankTbl.csv`) is an **ordered category** (`Unranked`, `Iron`, `Bronze`, `Silver`, `Gold`, `Platinum`, `Emerald`, `Diamond`, `Master`, `Grandmaster`, `Challenger`), while `Lane` is an **unordered category** (`TOP`, `JUNGLE`, `MIDDLE`, `BOTTOM`).
First, we must inspect **group sizes** using `.value_counts()`. Comparing a group with very few records to a massive group requires statistical caution.

**Predict first:**
* *Prediction for Group Sizes:* I expect the middle ranks (`Silver`, `Gold`, `Platinum`, `Emerald`) and `Unranked` to contain the vast majority of records, while apex tiers (`Master`, `Grandmaster`, `Challenger`) will have very small sample sizes.
* *Prediction for Boxplots across Lanes:* I predict that `BOTTOM` and `MIDDLE` will have the widest spread in `DmgDealt` and `MinionsKilled` (especially `BOTTOM`, which combines both ADC marksmen who farm minions and Supports who do not), whereas `TOP` and `MIDDLE` will have the highest median `MinionsKilled`.

```python
# 1. Sprawdzenie liczebności grup dla uporządkowanej kategorii RankFk / RankName
rank_order = rank_df.sort_values("RankId")["RankName"].tolist()
rank_counts = lol_data["RankName"].value_counts().reindex(rank_order, fill_value=0)
print("--- Record counts by Rank Tier (Ordered Category) ---")
print(rank_counts)

fig, ax = plt.subplots(figsize=(9, 4))
rank_counts.plot.bar(ax=ax, color="steelblue", rot=30)
ax.set_xlabel("Competitive Rank Tier (Ordered Category)")
ax.set_ylabel("Number of player records")
ax.set_title("Number of recorded match performances by Rank Tier")
plt.tight_layout()
plt.show()

# 2. Porównanie rozkładów DmgDealt oraz MinionsKilled względem linii (Lane) za pomocą Boxplota
valid_lanes = df_classic.loc[df_classic["Lane"].isin(["TOP", "JUNGLE", "MIDDLE", "BOTTOM"])].copy()
print("\n--- Record counts by Lane in CLASSIC mode ---")
print(valid_lanes["Lane"].value_counts())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

valid_lanes.boxplot(column="DmgDealt", by="Lane", ax=axes[0], grid=False)
axes[0].set_title("Damage Dealt distribution by Lane")
axes[0].set_xlabel("Assigned Lane")
axes[0].set_ylabel("Damage Dealt to Champions")

valid_lanes.boxplot(column="MinionsKilled", by="Lane", ax=axes[1], grid=False)
axes[1].set_title("Minions Killed (CS) distribution by Lane")
axes[1].set_xlabel("Assigned Lane")
axes[1].set_ylabel("Minions Killed (CS)")

fig.suptitle("")  # Usuwamy domyślny nadtytuł pandas
plt.tight_layout()
plt.show()

# 3. Boxplot visionScore lub TotalGold według uporządkowanej kategorii RankId (dla rang z wystarczającą próbą)
fig, ax = plt.subplots(figsize=(10, 4))
df_classic.boxplot(column="TotalGold", by="RankFk", ax=ax, grid=False)
fig.suptitle("")
ax.set_title("Total Gold Earned distribution across Ordered Rank IDs")
ax.set_xlabel("Rank ID (0 = Unranked, 1 = Iron -> 10 = Challenger)")
ax.set_ylabel("Total Gold Earned")
plt.tight_layout()
plt.show()
```

### Exercise 3 — Boxplot Interpretation, Group Sizes, and Critique

* **Comparison (Boxplot vs. Scatterplot):**
  Unlike a scatterplot where thousands of overlapping points obscure the middle of the data, the **boxplot** clearly displays the **median** (the horizontal line inside the box) and the **Interquartile Range (IQR = 75th percentile − 25th percentile)**, which contains the middle 50% of observations. For example, in `MinionsKilled` by `Lane`, the `BOTTOM` lane box has a huge IQR (height) because it merges two distinct roles: Marksmen (ADCs with 180+ CS) and Supports (with 20–40 CS, as seen in Rows 0–2 of `MatchStatsTbl.csv` where `MinionsKilled` is `30, 29, 34` and `visionScore` is `67, 88, 97`).
* **Group Sizes Caution:**
  When inspecting `RankName` counts, tiers at the extreme top (or specific unpopulated ranks) have far fewer observations than `Unranked` or mid-tier ranks. Claims about the median or IQR of a tiny group are statistically fragile because a single player's match history can heavily skew the entire box.
* **Critique of a False Claim:**
  *Suppose a student claims: "All points above the top whisker (1.5 × IQR) in the `DmgDealt` boxplot are cheaters or data errors, and players with low damage below the median played poorly."*
  **Why the plot does not support this:** Points beyond $Q_3 + 1.5 \times \text{IQR}$ are merely **statistical outliers** in a right-skewed distribution, not errors. A 50-minute match naturally produces 60,000+ damage. Conversely, a player with low damage to champions might have played a split-pushing strategy (focusing on `TurretDmgDealt`), a healing support (like Soraka in Row 2, who won the game with only `6,410` damage but `97` vision score!), or won an early 15-minute surrender.

## Exercise 4 — Filter records, create an efficiency measure, and rank a shortlist

**Scenario:** An esports analyst wants to study dominant **Top Lane (`Lane == "TOP"`)** performances in **winning games (`Win == 1`)** where the player demonstrated strong farming and survivability: **at least 180 minions killed (`MinionsKilled >= 180`)** and **at most 2 deaths (`deaths <= 2`)** (with `TotalGold > 0`).

### New Measure Definition:
We define **Damage per Gold (`dmg_per_gold`) = `DmgDealt / TotalGold`**, measured in champion damage dealt per 1 unit of gold earned.
* For example, a player who earned `15,000` gold and dealt `30,000` damage has a `dmg_per_gold` ratio of `2.00`.
* **Larger values indicate higher combat efficiency** (converting economic resources into maximum champion damage).
We also compute **KDA Ratio = `(kills + assists) / maximum(1, deaths)`** for additional context.

```python
# 1. Filtrowanie podzbioru za pomocą maski logicznej (Boolean mask) z zachowaniem wszystkich kolumn (.copy())
top_mask = (
    (lol_data["Lane"] == "TOP") &
    (lol_data["Win"] == 1) &
    (lol_data["MinionsKilled"] >= 180) &
    (lol_data["deaths"] <= 2) &
    (lol_data["TotalGold"] > 0)
)

matching_matches = lol_data.loc[top_mask].copy()

# Checkpoint weryfikujący warunki (jak w oryginalnym zadaniu)
assert len(matching_matches) > 0, "Selection is empty; check conditions."
assert (matching_matches["Lane"] == "TOP").all(), "Failed Lane condition."
assert (matching_matches["Win"] == 1).all(), "Failed Win condition."
assert (matching_matches["MinionsKilled"] >= 180).all(), "Failed MinionsKilled condition."
assert (matching_matches["deaths"] <= 2).all(), "Failed deaths condition."
print(f"Checkpoint passed! Number of matching TOP lane records: {len(matching_matches):,}")

# 2. Porównanie statystyk podzbioru z całym zbiorem danych
print("\n--- Summary of matching_matches (DmgDealt & TotalGold) ---")
display(matching_matches[["DmgDealt", "TotalGold", "MinionsKilled", "TurretDmgDealt"]].describe().round(2))

subset_median_dmg = matching_matches["DmgDealt"].median()
full_median_dmg = lol_data["DmgDealt"].median()
print(f"Median DmgDealt (Matching TOP Subset): {subset_median_dmg:,.2f}")
print(f"Median DmgDealt (Full Dataset):        {full_median_dmg:,.2f}")

# 3. Stworzenie nowych miar: dmg_per_gold oraz KDA
matching_matches["dmg_per_gold"] = matching_matches["DmgDealt"] / matching_matches["TotalGold"]
matching_matches["kda"] = (matching_matches["kills"] + matching_matches["assists"]) / matching_matches["deaths"].replace(0, 1)

# 4. Stworzenie shortlisty TOP 5 występów o najwyższej efektywności obrażeń na jednostkę złota (dmg_per_gold)
shortlist = matching_matches.sort_values("dmg_per_gold", ascending=False).head(5).copy()

display_cols = ["MatchStatsId", "MatchFk", "ChampionName", "DmgDealt", "TotalGold", "dmg_per_gold", "MinionsKilled", "TurretDmgDealt", "kills", "deaths", "assists", "kda"]
print("\n--- Shortlist: Top 5 Most Gold-Efficient TOP Lane Performances ---")
display(shortlist[display_cols].round(2))

# 5. Wykres wyróżniający podzbiór i shortlistę na tle wszystkich gier (szare -> niebieskie -> pomarańczowe obwódki)
bg_sample = lol_data.sample(n=min(20000, len(lol_data)), random_state=42)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(bg_sample["TotalGold"], bg_sample["DmgDealt"], color="grey", s=10, alpha=0.15, label="All records (sample)")
ax.scatter(matching_matches["TotalGold"], matching_matches["DmgDealt"], color="steelblue", s=20, alpha=0.5, label="Matching TOP wins (CS>=180, Deaths<=2)")
ax.scatter(shortlist["TotalGold"], shortlist["DmgDealt"], facecolors="none", edgecolors="darkorange", s=120, linewidths=2.2, label="Top 5 Shortlist (Highest Dmg/Gold)")

ax.set_xlabel("Total Gold Earned (Gold)")
ax.set_ylabel("Damage Dealt to Champions")
ax.set_title("Matching TOP Lane Performances and Damage-per-Gold Shortlist")
ax.ticklabel_format(style="plain", axis="both")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

### Exercise 4 — Interpretation & Limitations

* **Subset Summary:**
  Filtering for winning `TOP` lane games with `MinionsKilled >= 180` and `deaths <= 2` isolates high-performing laning games. The median `DmgDealt` and `TotalGold` in `matching_matches` are substantially higher than the full dataset's medians because the filter removes short games, supports, and losing performances where players spent significant time dead.
* **Trade-offs in the Shortlist:**
  The orange-outlined shortlist points lie along the upper-left/top boundary of the scatterplot — achieving massive `DmgDealt` relative to their `TotalGold`. However, inspecting `TurretDmgDealt` and `kda` in the shortlist table shows a classic strategic trade-off: players who maximize `dmg_per_gold` often play poke/AoE fighters (constantly brawling enemy champions) rather than dedicated split-pushers who focus on destroying structures (`TurretDmgDealt`).
* **Limitations of the Ratio (`dmg_per_gold`):**
  1. **What the ratio ignores:** `dmg_per_gold` completely ignores **objective control (`TurretDmgDealt`, `DragonKills`, `BaronKills`)**, **crowd control (CC)**, **vision (`visionScore`)**, and **damage tanked (`DmgTaken`)**. A top-laner who split-pushes and deals `12,000` turret damage to win the game might have a low `dmg_per_gold` ratio against champions, yet was the most valuable player in the match.
  2. **Matchup context:** Dealing high damage against a high-HP regeneration tank (e.g., Mundo, Zac, Cho'Gath) inflates `DmgDealt` without necessarily resulting in kills or map pressure.

## Final Reflection

1. **Two Key Findings Supported by Evidence:**
   * **Right-skewness and role-dependent variance:** In-game combat metrics (`DmgDealt`) are strongly right-skewed, making the **median** a much more reliable measure of a typical performance than the **mean**. Furthermore, boxplots across `Lane` show that role dictates distribution shape — `BOTTOM` exhibits an abnormally wide IQR in `MinionsKilled` because it combines farming Marksmen with non-farming Supports.
   * **Resource conversion vs. confounding variables:** While Pearson correlation and scatterplots show a strong positive relationship between `TotalGold` and `DmgDealt`, the vertical spread widens at high gold values, proving that gold efficiency (`dmg_per_gold`) depends heavily on champion class, game mode (`QueueType`), and match length (`GameDuration`).
2. **Dataset Limitations:**
   * **Sampling & Relational Mismatch:** Notice that `MatchStatsTbl.csv` has **732,308** player rows while `MatchTbl.csv` has **279,423** matches (in a complete dataset of 10 players per match, 279k matches would equal ~2.79 million player rows). This means `MatchStatsTbl.csv` only stores performances for tracked summoners (`SummonerFk`), not all 10 participants of every match. Additionally, the dataset mixes patches (`14.23`, `14.24`, etc.) and game modes (`ARAM`, `CLASSIC`), and suffers from a corrupted `CurrentMasteryPoints` column that duplicates `ChampionFk`.
